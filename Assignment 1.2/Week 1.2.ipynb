# Part 1: the Atom class

```python
class Atom:
    def __init__(self, symbol, atomic_num, num_neutrons):
        """
        Initializes an Atom object with symbol, atomic number, and number of neutrons.
        
        Args:
            symbol (str): The symbol of the element.
            atomic_num (int): The atomic number of the element.
            num_neutrons (int): The number of neutrons in the atom's nucleus.
        """
        self.symbol = symbol
        self.atomic_num = atomic_num
        self.num_neutrons = num_neutrons

    def proton_number(self):
        """
        Returns the number of protons (atomic number) in the nucleus of the atom.
        
        Returns:
            int: The number of protons in the atom.
        """
        return self.atomic_num
    
    def mass_number(self):
        """
        Calculates and returns the mass number of the atom.
        The mass number is the sum of the atomic number and the number of neutrons.
        
        Returns:
            int: The mass number of the atom.
        """
        return self.atomic_num + self.num_neutrons    

    def isotope(self, num_neutrons):
        """
        Changes the number of neutrons in the atom's nucleus to the specified value.
        
        Args:
            num_neutrons (int): The new number of neutrons in the atom's nucleus.
        """
        self.num_neutrons = num_neutrons

    
    def __gt__(self, other):
        """
        Checks if an isotope is greater than another isotope based on their symbol and mass number.
        
        Args:
            other (Atom): The other Atom object to compare with.
        
        Returns:
            bool: True if the isotope is greater than the other, False otherwise.
        
        Raises:
            ValueError: If the elements of the isotopes are different.
                Isotopes of different elements cannot be compared.
        """
        if isinstance(other, Atom) and self.symbol == other.symbol:
            return self.mass_number() > other.mass_number()
        raise ValueError("Isotopes of different elements cannot be compared.")  
      
    def __le__(self, other):
        """
        Checks if an isotope is less than or equal to another isotope based on their symbol and mass number.
        
        Args:
            other (Atom): The other Atom object to compare with.
        
        Returns:
            bool: True if the isotope is less than or equal to the other, False otherwise.
        
        Raises:
            ValueError: If the elements of the isotopes are different.
                Isotopes of different elements cannot be compared.
        """
        if isinstance(other, Atom) and self.symbol == other.symbol:
            return self.mass_number() <= other.mass_number()
        raise ValueError("Isotopes of different elements cannot be compared.")
```

```python
# Testing the code
protium = Atom('H', 1, 1)
deuterium = Atom('H', 1, 2)
oxygen = Atom('O', 8, 8)
tritium = Atom('H', 1, 2)
tritium.isotope(3)

assert tritium.num_neutrons == 3
assert tritium.mass_number() == 4
assert protium < deuterium
assert deuterium <= tritium
assert tritium >= protium
print (oxygen > tritium) # <-- this should raise an Exception
```

# Part 2: the Molecule class

```python
class Molecule:
    def __init__(self, atom_pairs):
        """
        Initializes a Molecule object with a list of atom pairs.
        
        Args:
            atom_pairs (list): List of tuples containing Atom objects and their counts in the molecule.
                Each tuple has two elements: the Atom object and the number of atoms of that type.
        """
        self.atom_pairs = atom_pairs

    def count_atoms(self):
        """
        Counts the total number of atoms in the molecule.
        
        Returns:
            int: The total number of atoms in the molecule.
        """
        total_atoms = 0
        for atom, count in self.atom_pairs:
            total_atoms += count
        return total_atoms
    
    def __str__(self):
        """
        Returns a string representation of the molecule in the form of a chemical formula.
        Omitting the number 1 for individual atoms.
        """
        formula = ""
        for atom, count in self.atom_pairs:
            if count > 1:
                formula += atom.symbol + str(count)
            else:
                formula += atom.symbol
        return formula

    def __add__(self, other):
        """
        Adds two Molecule objects to create a new Molecule object.
        
        Args:
            other (Molecule): The other Molecule object to add.
        
        Returns:
            Molecule: A new Molecule object representing the combined molecule.
        """
        new_atom_pairs = self.atom_pairs + other.atom_pairs
        return Molecule(new_atom_pairs)
```

```python
# Testing the code
hydrogen = Atom('H', 1, 1)
carbon = Atom('C', 6, 6)
oxygen = Atom('O', 8, 8)

water = Molecule( [ (hydrogen, 2), (oxygen, 1) ] )
co2 = Molecule( [ (carbon, 1), (oxygen, 2) ])
print (water) # H2O
print (co2) # CO2
print (water + co2) # H2OCO2
```

# Part 3: The Chloroplast class

```python
class Chloroplast:
    def __init__(self):
        """
        Initializes a Chloroplast object with water and CO2 fields set to 0.
        """
        self.water = 0
        self.co2 = 0

    def add_molecule(self, molecule):
        """
        Adds a molecule to the chloroplast, incrementing the corresponding field.
        Performs photosynthesis when the total count of water and CO2 molecules reaches a certain threshold.

        Args:
            molecule (Atom): The molecule to add (Atom object representing either 'H2O' or 'CO2').

        Returns:
            list: A list of tuples representing the resulting molecules after photosynthesis.

        Raises:
            ValueError: If the molecule is not 'H2O' or 'CO2'.
        """
        if str(molecule) == 'H2O':
            self.water += 1
        elif str(molecule) == 'CO2':
            self.co2 += 1
        else:
            raise ValueError("You can only add water or CO2 molecules")

        if self.water >= 12 and self.co2 >= 6:
            self.water -= 12
            self.co2 -= 6
            # Molecule of sugar and molecules of oxygen
            return [("C6H12O6", 1), ("O2", 6)]
        else:
            return []

    def __str__(self):
        """
        Returns a string representation of the Chloroplast object,
        displaying the number of water and CO2 molecules stored.

        Returns:
            str: A string representation of the Chloroplast object.
        """
        return f"Stored molecules: {self.water} water and {self.co2} CO2"
```
