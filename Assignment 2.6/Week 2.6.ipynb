# The article 'Software Engineering for Machine Learning: Characterizing and Detecting Mismatch in Machine-Learning Systems' by Lewis and Ozkaya (2021)

The article discusses the challenges in developing and deploying ML-enabled systems and introduces the concept of ML mismatches, which occur when there are incorrect assumptions between different components of the system. The authors identify seven categories of ML mismatches:

1. Trained model: Mismatches in this category include a lack of information about test cases, test data, and API specifications. For example, if a data scientist fails to provide a clear description of the model's components and specifications, it becomes challenging for software engineers to integrate the model into the system effectively.

2. Operational environment: Mismatches here relate to runtime metrics and data. If the operations staff is unaware of what metrics to monitor or fails to collect the necessary runtime data, it can affect the performance of the ML component in the production environment.

3. Task and purpose: These mismatches involve unclear business goals and requirements. Insufficient communication between project owners and data scientists can lead to inefficiencies. For instance, if the task given to a data scientist lacks proper specifications, it becomes difficult for them to deliver the expected results.

4. Raw data: Mismatches in this category arise from the absence of metadata and descriptions of data elements. Without information about how the data was collected, its distribution, and the meaning of missing values, data scientists face challenges in understanding and processing the data effectively.

5. Development environment: These mismatches often stem from differences in programming languages. If data scientists and software engineers fail to share information about the programming languages used, it can lead to compatibility issues. For example, attempting to reproduce a model developed in one language (e.g., R) using a different language (e.g., Python) can result in errors due to language-specific differences.

6. Operational data: Mismatches in this category result from a lack of statistics on operational data. If the data used to train the model does not accurately represent the data in the production environment, the model's performance may be suboptimal.

7. Training data: Mismatches here arise from a lack of details about data preparation pipelines. Without information on how the training data was derived from raw data, data scientists may face limitations in exploring alternative feature engineering techniques.

The article highlights the importance of improved communication and the development of automated tools to detect and prevent these mismatches. The authors aim to create machine-readable descriptors that capture the necessary information to avoid mismatches and facilitate collaboration among stakeholders in ML-enabled system development. 

# The article 'Tackling Collaboration Challenges in the Development of ML-Enabled Systems' by Lewis (2023)

In this article, the author discusses the challenges of collaboration in the development of ML-enabled systems and presents findings from a study on this topic. The study was conducted through interviews with professionals building ML-enabled systems and a literature review on software engineering for ML-enabled systems. The three main collaboration points identified are requirements and planning, training data, and product-model integration.

1. **Requirements and Planning**: One of the challenges is the tension between product requirements and model requirements. In some organizations, the model is built first and then the product is developed around it, while in others, the product is developed first, and then a model is built to support it. This can lead to misunderstandings and unrealistic expectations about model capabilities.

Recommendations:
- Involve data scientists early in the process to better understand ML capabilities.
- Consider adopting a parallel development trajectory for product and model teams. 

2. **Training Data**: Disagreements over training data represent the most common collaboration challenges. Model teams often do not own, collect, or fully understand the data they need, leading to data quality and quantity issues.

Recommendations:
- Plan and budget for data collection and access to domain experts.
- Adopt a formal contract that specifies data quality and quantity expectations.
- Employ a data validation and monitoring infrastructure early in the project. 

3. **Product-Model Integration**: This collaboration point involves close cooperation between data scientists and software engineers. Conflicts can arise due to differences in cultures, unclear processes and responsibilities, and different quality assurance approaches.

Recommendations:
- Define processes, responsibilities, and boundaries more carefully. 
- Encourage open communication and avoid siloing data scientists. 

4. **Quality Assurance for Model and Product**: Ensuring model adequacy and transparent evaluation, as well as responsibility for system testing, are challenges in this collaboration point.

Recommendations:
- Plan for structured feedback from the product engineering team to the model team. 
- Define clear quality requirements for model and product.

The recommendations provided in the study can help teams navigate these challenges and work harmoniously to build successful ML-enabled systems.
